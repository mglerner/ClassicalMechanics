**Goal:** In this notebook, we'll visualize the solutions to the 2D harmonic oscillator. Our goal is to understand how the particle's trajectory (its path in the $xy$-plane) depends on the amplitudes, frequencies, and the relative phase shift between the $x$ and $y$ motions.

## 1. Setup

Run this cell once. It defines $x(t)$ and $y(t)$, the menus of values you can pick from, and the plot.

```python
%matplotlib widget
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from ipywidgets import interact

sns.set_theme()          # seaborn's default look

def x(t, Ax, omega):
    return Ax * np.cos(omega * t)

def y(t, Ay, omega, delta):
    return Ay * np.cos(omega * t - delta)

# the menus Will used
pi = np.pi
amplitudes = [0.5, 1, 2]
omega_menu = [("pi", pi), ("2 pi", 2 * pi), ("4 pi", 4 * pi)]
omega_y_menu = [("pi", pi), ("sqrt(2 pi)", np.sqrt(2 * pi)), ("2 pi", 2 * pi), ("3 pi", 3 * pi),
                ("4 pi", 4 * pi), ("6 pi", 6 * pi), ("10 pi", 10 * pi)]

def new_plot():
    # one figure, made once; the sliders only move the two lines on it (a new figure per move flickers)
    plt.ioff()
    fig, ax = plt.subplots(figsize=(5, 5))
    plt.ion()
    fig.canvas.header_visible = False
    path, = ax.plot([], [])
    dot, = ax.plot([], [], "o")          # the particle at t = tmax
    ax.set(xlim=(-2.1, 2.1), ylim=(-2.1, 2.1), xlabel="x", ylabel="y", aspect="equal")
    return fig, path, dot

def draw(path, dot, tmax, Ax, Ay, omega_x, omega_y, delta):
    t = np.linspace(0, tmax, 2000)
    path.set_data(x(t, Ax, omega_x), y(t, Ay, omega_y, delta))
    dot.set_data([x(tmax, Ax, omega_x)], [y(tmax, Ay, omega_y, delta)])
    path.figure.canvas.draw_idle()
```

## 2. Part 1: The Isotropic Oscillator ($\omega_x = \omega_y = \omega$)

First, we'll explore the case where the restoring force is the same in both directions, meaning the oscillation frequency $\omega$ is identical for both $x(t)$ and $y(t)$.

The solutions are: $x(t) = A_x\cos(\omega t)$ and $y(t) = A_y\cos(\omega t - \delta)$

Run the following block of code (tmax is in seconds):

```python
fig1, path1, dot1 = new_plot()

def isotropic(tmax=0.01, Ax=1, Ay=1, omega=2 * pi, delta=0.0):
    draw(path1, dot1, tmax, Ax, Ay, omega, omega, delta)

interact(isotropic, tmax=(0.01, 2.0, 0.01), Ax=amplitudes, Ay=amplitudes, omega=omega_menu, delta=(0.0, 2 * pi, 0.01));
fig1.canvas
```

## 3. Part 2: The Anisotropic Oscillator ($\omega_x \ne \omega_y$)

Now we'll explore the more complex patterns that emerge when the frequencies are different. These beautiful paths are known as **Lissajous figures**. The key factor determining the shape is the ratio of the two frequencies, $\omega_x/\omega_y$. Run the code cell below. Notice that $\omega_x$ and $\omega_y$ can now be controlled independently.

The solutions are $x(t) = A_x\cos(\omega_x t)$ and $y(t) = A_y\cos(\omega_y t - \delta)$

```python
fig2, path2, dot2 = new_plot()

def anisotropic(tmax=0.01, Ax=1, Ay=1, omega_x=2 * pi, omega_y=2 * pi, delta=pi / 2):
    draw(path2, dot2, tmax, Ax, Ay, omega_x, omega_y, delta)

interact(anisotropic, tmax=(0.01, 4.0, 0.01), Ax=amplitudes, Ay=amplitudes, omega_x=omega_menu, omega_y=omega_y_menu,
         delta=(0.0, 2 * pi, 0.01));
fig2.canvas
```

## 4. Extra: watch it move

Optional. The same plot as Part 2, with a play button that advances $t_{\max}$ for you (in steps of 0.01 s). Everything else is as above.

```python
from ipywidgets import FloatSlider, Play, HBox, VBox, interactive_output

fig3, path3, dot3 = new_plot()
tmax3 = FloatSlider(value=0.01, min=0.01, max=4, step=0.01, description="tmax (s)")
play = Play(min=1, max=400, step=1, interval=40, description="play")

def advance(change):
    tmax3.value = change["new"] / 100
play.observe(advance, "value")

def animated(tmax, Ax, Ay, omega_x, omega_y, delta):
    draw(path3, dot3, tmax, Ax, Ay, omega_x, omega_y, delta)

controls = interact(animated, tmax=tmax3, Ax=amplitudes, Ay=amplitudes, omega_x=omega_menu, omega_y=omega_y_menu,
                    delta=(0.0, 2 * pi, 0.01)).widget
VBox([HBox([play, tmax3]), controls, fig3.canvas])
```
