## 2. Setup

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme()          # seaborn's default look for every plot below
from ipywidgets import interact, FloatSlider, Dropdown, Play, HBox

def x(t, Ax, omega):
    return Ax * np.cos(omega * t)

def y(t, Ay, omega, delta):
    return Ay * np.cos(omega * t - delta)

def plot_path(tmax, Ax, Ay, omega_x, omega_y, delta):
    t = np.linspace(0, tmax, 2000)
    plt.figure(figsize=(5, 5))
    plt.plot(x(t, Ax, omega_x), y(t, Ay, omega_y, delta))
    plt.plot(x(tmax, Ax, omega_x), y(tmax, Ay, omega_y, delta), "o")   # the particle at t = tmax
    plt.xlim(-2.1, 2.1)
    plt.ylim(-2.1, 2.1)
    plt.gca().set_aspect("equal")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.show()

# the menus Will used
amplitudes = [0.5, 1, 2]
pi = np.pi
omega_menu = [("pi", pi), ("2 pi", 2 * pi), ("4 pi", 4 * pi)]
omega_y_menu = [("pi", pi), ("sqrt(2 pi)", np.sqrt(2 * pi)), ("2 pi", 2 * pi), ("3 pi", 3 * pi),
                ("4 pi", 4 * pi), ("6 pi", 6 * pi), ("10 pi", 10 * pi)]

def play_button(tmax_slider):
    # a play button that drives the tmax slider, in steps of 0.01 s
    play = Play(min=1, max=int(round(tmax_slider.max * 100)), step=1, interval=40, description="play")
    def step(change):
        tmax_slider.value = change["new"] / 100
    play.observe(step, "value")
    return play
```

## 3. Part 1: The Isotropic Oscillator ($\omega_x = \omega_y = \omega$)

First, we'll explore the case where the restoring force is the same in both directions, meaning the oscillation frequency $\omega$ is identical for both $x(t)$ and $y(t)$.

The solutions are: $x(t) = A_x\cos(\omega t)$ and $y(t) = A_y\cos(\omega t - \delta)$

Run the following block of code:

```python
def isotropic(tmax, Ax, Ay, omega, delta):
    plot_path(tmax, Ax, Ay, omega, omega, delta)

tmax1 = FloatSlider(value=0.01, min=0.01, max=2, step=0.01, description="tmax (s)")
display(HBox([play_button(tmax1)]))
interact(isotropic, tmax=tmax1, Ax=Dropdown(options=amplitudes, value=1), Ay=Dropdown(options=amplitudes, value=1),
         omega=Dropdown(options=omega_menu, value=2 * pi),
         delta=FloatSlider(value=0, min=0, max=2 * pi, step=0.01, description="delta"));
```

## 4. Part 2: The Anisotropic Oscillator ($\omega_x \ne \omega_y$)

Now we'll explore the more complex patterns that emerge when the frequencies are different. These beautiful paths are known as **Lissajous figures**. The key factor determining the shape is the ratio of the two frequencies, $\omega_x/\omega_y$. Run the code cell below. Notice that $\omega_x$ and $\omega_y$ can now be controlled independently.

The solutions are $x(t) = A_x\cos(\omega_x t)$ and $y(t) = A_y\cos(\omega_y t - \delta)$

```python
tmax2 = FloatSlider(value=0.01, min=0.01, max=4, step=0.01, description="tmax (s)")
display(HBox([play_button(tmax2)]))
interact(plot_path, tmax=tmax2, Ax=Dropdown(options=amplitudes, value=1), Ay=Dropdown(options=amplitudes, value=1),
         omega_x=Dropdown(options=omega_menu, value=2 * pi), omega_y=Dropdown(options=omega_y_menu, value=2 * pi),
         delta=FloatSlider(value=pi / 2, min=0, max=2 * pi, step=0.01, description="delta"));
```
