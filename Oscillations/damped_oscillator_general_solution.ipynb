**General solution:** $\;x(t) = e^{-\beta t}\left(C_1\, e^{\sqrt{\beta^2 - \omega_0^2}\,t} + C_2\, e^{-\sqrt{\beta^2 - \omega_0^2}\,t}\right)$


```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact

t = np.linspace(0, 20, 1000)

def plot_damped(beta=0.0, omega0=1.0, C1=1.0, C2=1.0):
    # imaginary when beta < omega0 (the oscillation)
    q = np.sqrt(complex(beta**2 - omega0**2))        
    x = np.exp(-beta * t) * (C1 * np.exp(q * t) + C2 * np.exp(-q * t))
    plt.figure(figsize=(8, 4))
    # the real part (for beta < omega0 the two terms are complex conjugates when C1 = C2)
    plt.plot(t, x.real, 'k')                        
    if beta < omega0:
        # envelope: the decaying amplitude
        plt.plot(t, (C1 + C2) * np.exp(-beta * t), 'r--')
        plt.plot(t, -(C1 + C2) * np.exp(-beta * t), 'r--')
    else:
        # no oscillation: show the two pieces instead
        plt.plot(t, (C1 * np.exp(-beta * t) * np.exp(q * t)).real, 'r--')
        plt.plot(t, (C2 * np.exp(-beta * t) * np.exp(-q * t)).real, 'b--')
    qs = f'{q.imag:.2f}i' if q.real == 0 else f'{q.real:.2f}'
    plt.title(fr'$x(t) = e^{{-{beta:.2f}\,t}}\,({C1:.1f}\,e^{{{qs}\,t}} + {C2:.1f}\,e^{{-{qs}\,t}})$')
    plt.xlabel('t (s)'); plt.ylabel('x')
    plt.ylim(-2.5, 2.5)
    plt.show()

interact(plot_damped, beta=(0.0, 3.0, 0.01), 
         omega0=(0.1, 3.0, 0.1), 
         C1=(-2.0, 2.0, 0.1), C2=(-2.0, 2.0, 0.1));
```

## Interesting cases

With $\omega_0 = 1$ throughout:

- **Undamped:** $\beta = 0$, $C_1 = C_2 = 1$. A pure cosine, $x = 2\cos t$; the envelope is flat.
- **Underdamped:** $\beta = 0.1$, $C_1 = C_2 = 1$. Many oscillations inside a slowly shrinking envelope.
  $\beta = 0.5$: only a few swings before it is gone. The frequency inside the envelope,
  $\omega_1 = \sqrt{\omega_0^2 - \beta^2}$, is slightly less than $\omega_0$ (the exponent in the title).
- **Critically damped:** $\beta = 1$. The square root is zero, the two exponentials collapse into one,
  and W15's formula only gives $x = (C_1 + C_2)\,e^{-t}$: the fastest return with no overshoot, but the formula
  has lost a solution. The full solution here is $x = (A + Bt)\,e^{-\beta t}$ (Taylor 5.4; Will's W19).
- **Overdamped:** $\beta = 2$, $C_1 = C_2 = 1$. No oscillation; the two dashed pieces decay at different rates
  and the slow one, $e^{-(\beta - \sqrt{\beta^2 - \omega_0^2})\,t}$, is what you see at late times.
- **Pushed from the origin:** $\beta = 2$, $C_1 = 1$, $C_2 = -1$. Starts at $x = 0$ with a positive velocity,
  rises, then creeps back (Taylor's overdamped figure). $x(0) = C_1 + C_2$ and $v(0) = \sqrt{\beta^2 - \omega_0^2}\,(C_1 - C_2)$.
