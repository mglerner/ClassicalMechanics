**General solution:** $\;x(t) = e^{-\beta t}\left(C_1\, e^{\sqrt{\beta^2 - \omega_0^2}\,t} + C_2\, e^{-\sqrt{\beta^2 - \omega_0^2}\,t}\right)$


```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact

t = np.linspace(0, 20, 1000)

def plot_damped(beta=0.0, omega0=1.0, C1=1.0, C2=1.0):
    q = np.sqrt(complex(beta**2 - omega0**2))        # imaginary when beta < omega0: that is the oscillation
    x = np.exp(-beta * t) * (C1 * np.exp(q * t) + C2 * np.exp(-q * t))
    plt.figure(figsize=(8, 4))
    plt.plot(t, x.real, 'k')                        # the real part (for beta < omega0 the two terms are complex conjugates when C1 = C2)
    if beta < omega0:                               # envelope: the decaying amplitude
        plt.plot(t, (C1 + C2) * np.exp(-beta * t), 'r--')
        plt.plot(t, -(C1 + C2) * np.exp(-beta * t), 'r--')
    else:                                           # no oscillation: show the two pieces instead
        plt.plot(t, (C1 * np.exp(-beta * t) * np.exp(q * t)).real, 'r--')
        plt.plot(t, (C2 * np.exp(-beta * t) * np.exp(-q * t)).real, 'b--')
    qs = f'{q.imag:.2f}i' if q.real == 0 else f'{q.real:.2f}'
    plt.title(f'$x(t) = e^{{-{beta:.2f}\\,t}}\\,({C1:.1f}\\,e^{{{qs}\\,t}} + {C2:.1f}\\,e^{{-{qs}\\,t}})$')
    plt.xlabel('t (s)'); plt.ylabel('x')
    plt.ylim(-2.5, 2.5)
    plt.show()

interact(plot_damped, beta=(0.0, 2.0, 0.05), omega0=(0.1, 3.0, 0.1), C1=(-2.0, 2.0, 0.1), C2=(-2.0, 2.0, 0.1));
```
