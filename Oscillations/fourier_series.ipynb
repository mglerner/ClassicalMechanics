**Fourier series of a periodic function** (Taylor 5.7). Any function with period $\tau$ is a sum of cosines and sines at multiples of $\omega = 2\pi/\tau$:

$$f(t) = a_0 + \sum_{n=1}^{\infty}\left[a_n\cos(n\omega t) + b_n\sin(n\omega t)\right],$$

$$a_0 = \frac{1}{\tau}\int_{-\tau/2}^{\tau/2} f\,dt, \qquad a_n = \frac{2}{\tau}\int_{-\tau/2}^{\tau/2} f\cos(n\omega t)\,dt, \qquad b_n = \frac{2}{\tau}\int_{-\tau/2}^{\tau/2} f\sin(n\omega t)\,dt.$$

**Example:** $f(t) = 3t^2$ on $-3/2 < t < 3/2$, repeated with period $\tau = 3$. It is even, so every $b_n = 0$, and

$$a_0 = \frac{9}{4}, \qquad a_n = \frac{27\,(-1)^n}{(n\pi)^2}.$$

The slider $n_{max}$ is where the sum stops. Black: $f(t)$. Red: the partial sum. Thin dashed: the separate terms $a_0$ and $a_n\cos(n\omega t)$ being added up.


```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact

tau = 3.0
omega = 2 * np.pi / tau
t = np.linspace(-1.2 * tau, 1.2 * tau, 1000)

def f(t):
    # 3 t^2, repeated every tau
    s = (t + tau / 2) % tau - tau / 2
    return 3 * s**2

a0 = 9 / 4

def a(n):
    return 27 * (-1)**n / (n * np.pi)**2

def plot_series(nmax=1):
    plt.figure(figsize=(8, 4))
    total = a0 * np.ones_like(t)
    plt.plot(t, total, '--', lw=1)
    for n in range(1, nmax + 1):
        term = a(n) * np.cos(n * omega * t)
        plt.plot(t, term, '--', lw=1)
        total = total + term
    plt.plot(t, f(t), 'k', lw=2)
    plt.plot(t, total, 'r', lw=2)
    plt.title(fr'$n_{{max}} = {nmax}$')
    plt.xlabel('t'); plt.ylabel('f(t)')
    plt.ylim(-4, 8)
    plt.show()

interact(plot_series, nmax=(1, 10, 1));
```

**Example 5.4, the rectangular pulse** (Taylor Figures 5.22 and 5.23). Height $f_{max} = 2$ for a time $\Delta\tau = 0.25$, then zero, repeated with period $\tau = 1$. Also even, so $b_n = 0$, and

$$a_0 = \frac{f_{max}\,\Delta\tau}{\tau}, \qquad a_n = \frac{2 f_{max}}{n\pi}\sin\!\left(\frac{n\pi\,\Delta\tau}{\tau}\right) \quad (5.86),\ (5.87).$$

(Taylor takes $f_{max} = 1$; with 2 every curve is twice his.) The slider $N$ is the last harmonic kept: $a_0 + \sum_{n=1}^{N} a_n\cos(n\omega t)$.


```python
T = 1.0
dT = 0.25
fmax = 2.0
w = 2 * np.pi / T
tp = np.linspace(-2.5, 2.5, 2000)

def pulse(t):
    s = (t + T / 2) % T - T / 2
    return np.where(abs(s) <= dT / 2, fmax, 0.0)

def plot_pulse(N=1):
    total = fmax * dT / T * np.ones_like(tp)
    for n in range(1, N + 1):
        an = 2 * fmax / (n * np.pi) * np.sin(n * np.pi * dT / T)
        total = total + an * np.cos(n * w * tp)
    plt.figure(figsize=(8, 4))
    plt.plot(tp, pulse(tp), 'k', lw=1.5)
    plt.plot(tp, total, 'r', lw=2)
    plt.title(f'N = {N}')
    plt.xlabel('t'); plt.ylabel('f(t)')
    plt.ylim(-0.3, 2.3)
    plt.show()

interact(plot_pulse, N=(1, 20, 1));
```

## Interesting cases

- **$3t^2$, $n_{max} = 1$:** the constant $a_0 = 2.25$ (the average of $f$) plus one cosine already follows the parabola.
  The coefficients are $a_1 = -2.74$, $a_2 = +0.684$, $a_3 = -0.304$: they fall like $1/n^2$.
- **$3t^2$, $n_{max} = 10$:** hard to tell from $f$, except at the corners $t = \pm 3/2$, where $f$ has a kink.
- **Pulse, $N = 2$:** Taylor's Figure 5.23(a), "3 terms" (he counts $a_0$ as a term). The general shape, no corners.
- **Pulse, $N = 10$:** Figure 5.23(b), "11 terms". Close, with a small overshoot at each jump (the Gibbs phenomenon).
- **Pulse, $N = 4$ and $N = 5$:** the same curve. $a_4 = 0$ because $\sin(4\pi\,\Delta\tau/\tau) = \sin\pi = 0$; so are $a_8, a_{12}, \ldots$
- **Pulse, $N = 20$:** the overshoot at the jumps does not shrink as $N$ grows; it only gets narrower. The $a_n$ fall like $1/n$, slower than $3t^2$'s $1/n^2$: a jump costs more terms than a kink.
