# Two ways to write the same oscillation

$$x(t) = B_1 \cos\omega t + B_2 \sin\omega t \qquad\text{and}\qquad x(t) = A\cos(\omega t - \delta)$$

**From $A, \delta$ to $B_1, B_2$:** expand the cosine of a difference,
$$A\cos(\omega t - \delta) = A\cos\delta\,\cos\omega t + A\sin\delta\,\sin\omega t,$$
so matching the two forms term by term: $\;B_1 = A\cos\delta, \qquad B_2 = A\sin\delta.$

**From $B_1, B_2$ to $A, \delta$:** square and add, then divide:
$$A = \sqrt{B_1^2 + B_2^2}, \qquad \tan\delta = \frac{B_2}{B_1} \quad (\delta = \operatorname{atan2}(B_2, B_1),\ \text{which gets the quadrant right}).$$

**What the constants mean physically:** at $t = 0$,
$$x(0) = B_1, \qquad v(0) = \dot x(0) = \omega B_2 .$$
So $B_1$ is the starting position and $B_2$ is the starting velocity divided by $\omega$. A pure cosine ($B_2 = 0$, $\delta = 0$) starts at rest at its maximum; a pure sine ($B_1 = 0$, $\delta = \pi/2$) starts at the origin moving at full speed. The phase $\delta$ says how much of the motion is "sine-like", i.e. how much of the energy started out as velocity.


```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact

t = np.linspace(0, 10, 1000)

def plot_phase(B1=1.0, B2=0.0, omega=1.0):
    A = np.sqrt(B1**2 + B2**2)
    delta = np.arctan2(B2, B1)
    x0, v0 = B1, omega * B2                          # the initial position and velocity
    fig, (left, right) = plt.subplots(1, 2, figsize=(11, 4), gridspec_kw={'width_ratios': [1, 2.2]})
    # left: where it starts and how fast it is moving
    left.axhline(0, color='gray'); left.axvline(0, color='gray')
    left.plot([x0], [0], 'ko', markersize=10)
    left.annotate('', xy=(x0 + v0, 0), xytext=(x0, 0), arrowprops=dict(arrowstyle='->', color='red', lw=2))
    left.set_xlim(-3, 3); left.set_ylim(-1, 1); left.set_yticks([])
    left.set_xlabel('x'); left.set_title(f'at t = 0:  x = {x0:.2f},  v = {v0:.2f}')
    # right: the motion, both forms drawn on top of each other
    right.plot(t, B1 * np.cos(omega * t) + B2 * np.sin(omega * t), 'k', lw=3, label='$B_1\\cos\\omega t + B_2\\sin\\omega t$')
    right.plot(t, A * np.cos(omega * t - delta), 'y--', lw=2, label='$A\\cos(\\omega t - \\delta)$')
    right.plot([0], [x0], 'ko', markersize=10)
    right.axhline(0, color='gray')
    right.set_ylim(-3, 3); right.set_xlabel('t (s)'); right.set_ylabel('x')
    right.legend(loc='upper right')
    right.set_title(f'{B1:.2f} cos({omega:.1f} t) + {B2:.2f} sin({omega:.1f} t)  =  {A:.2f} cos({omega:.1f} t - {delta:.2f})')
    plt.show()

interact(plot_phase, B1=(-2.0, 2.0, 0.1), B2=(-2.0, 2.0, 0.1), omega=(0.5, 3.0, 0.1));
```
