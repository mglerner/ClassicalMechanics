# Kepler orbits

A planet or a comet moving under the sun's gravity follows the path
(Taylor Eq. 8.49, with the phase $\delta$ of Eq. 8.46 kept)

$$r(\phi) = \frac{c}{1 + \epsilon\cos(\phi - \delta)}, \qquad c = \frac{\ell^2}{\gamma\mu} .$$

The sun is at the origin and $c = 1$ here, so every distance is in units of $c$.
The sliders set the eccentricity $\epsilon$, how much of the path is drawn
($\phi$ from 0 to $\phi_\text{max}$), and the phase $\delta$. A port of Will
Raven's Mathematica notebook to Python.

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact

c = 1.0

def r_of_phi(phi, eps, delta):
    return c / (1 + eps * np.cos(phi - delta))

def draw(eps=0.5, phi_max=2 * np.pi, delta=0.0):
    phi = np.linspace(0, phi_max, 2000)
    r = r_of_phi(phi, eps, delta)
    # for eps >= 1 the formula gives r < 0 over part of the circle:
    # no orbit there, so leave a gap
    r[r < 0] = np.nan
    r[r > 50] = np.nan
    x = r * np.cos(phi)
    y = r * np.sin(phi)

    fig, ax = plt.subplots(figsize=(6, 6))
    ax.plot(x, y, color='tab:blue', lw=2)
    ax.plot(0, 0, 'o', color='orange', ms=10)      # the sun
    ax.set_xlim(-4, 4)
    ax.set_ylim(-4, 4)
    ax.set_aspect('equal')
    ax.grid(True, alpha=0.3)
    ax.set_xlabel(r'$x / c$')
    ax.set_ylabel(r'$y / c$')
    ax.set_title(fr'$\epsilon$ = {eps:.2f}')
    plt.show()

interact(
    draw,
    eps=(0.0, 1.5, 0.01),
    phi_max=(0.1, 2 * np.pi, 0.01),
    delta=(0.0, 2 * np.pi, 0.01),
);
```

## Interesting cases

- **A circle:** $\epsilon = 0$. The radius is $c$.
- **An ellipse:** $0 < \epsilon < 1$. Where is the sun: at the center, or somewhere
  else? Read $r_\text{min}$ and $r_\text{max}$ off the picture at $\epsilon = 0.5$
  and compare them with $c/(1 + \epsilon)$ and $c/(1 - \epsilon)$.
- **Halley's comet:** $\epsilon = 0.97$ (Taylor Example 8.4 has 0.967). Most of the
  orbit is off the picture.
- **A parabola:** $\epsilon = 1$. How close does it come to the sun, compared with
  the circle's radius?
- **A hyperbola:** $\epsilon > 1$. The path comes in, swings round the sun and leaves
  for good.
- **Which way the orbit points:** move $\delta$. What changes, and what does not?
- **Drawing it out:** set $\phi_\text{max}$ small and drag it up to $2\pi$.
