## 2. Setup

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import quad
```

## 3. `quad` on an integral you can do by hand

We know $\int_0^\pi \sin x\, dx = 2$. Here is how SciPy's `quad` does it:

```python
def integrand_sin(x):
    return np.sin(x)

value, error = quad(integrand_sin, 0, np.pi)
print(value)
print(error)
```

Line by line:

- `integrand_sin` is the function under the integral sign. `quad` wants a *function*, not a formula: you hand it the function's name, with no parentheses.
- `quad(integrand_sin, 0, np.pi)` integrates it from $x = 0$ to $x = \pi$.
- `quad` hands back **two** numbers. The first is the integral. The second is `quad`'s own estimate of how far off the first one might be (an absolute error). Writing `value, error = ...` catches both.
- Check the second number every time: it should be tiny compared with the first. Here it is around $10^{-14}$ against an answer of 2.

## 4. An integral along a path

Problem 6.23's integral runs along a path $y(x)$ and uses the path's slope $y'(x)$. The arc length of a path has the same shape:
$$L = \int_0^1 \sqrt{1 + y'^2}\, dx .$$
For the parabola $y = x^2$ from $x = 0$ to $x = 1$ this one can be done by hand: $L = \left[2\sqrt5 + \sinh^{-1} 2\right]/4 = 1.4789\ldots$, so we can check `quad` against it.

```python
def y_demo(x):
    return x**2

def y_demo_prime(x):
    return 2 * x

def length_integrand(x):
    return np.sqrt(1 + y_demo_prime(x)**2)

length, error = quad(length_integrand, 0, 1)
exact = (2 * np.sqrt(5) + np.arcsinh(2)) / 4
print(f"quad:  {length:.6f}  (error estimate {error:.1e})")
print(f"exact: {exact:.6f}")
```

Line by line:

- One short function for the path, one for its slope, and one for the integrand. The integrand calls the slope function. (Arc length needs only the slope; the 6.23 integrand needs both $y$ and $y'$, so yours will call both.)
- `f"..."` is a formatted string: whatever is inside `{ }` gets computed and printed. `:.6f` means "six digits after the decimal point"; `:.1e` means "scientific notation, one digit after the point".
- The two printed numbers agree, so the setup is right. That comparison with an answer you trust is the habit to keep.

It also helps to look at the path. `np.linspace(0, 1, 200)` makes 200 evenly spaced $x$ values, and `y_demo` works on all of them at once:

```python
x = np.linspace(0, 1, 200)
plt.figure(figsize=(6, 3.5))
plt.plot(x, y_demo(x))
plt.xlabel("x")
plt.ylabel("y")
plt.title("the path y = x^2")
plt.grid(alpha=0.3)
plt.show()
```
