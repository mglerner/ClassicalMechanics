**The brachistochrone is a cycloid** (Taylor 6.3, Example 6.2). With $y$ measured DOWN from point 1, the fastest track from rest is

$$x = a(\theta - \sin\theta), \qquad y = a(1 - \cos\theta) \qquad (6.26)$$

the curve traced by a point on the rim of a wheel of radius $a$ rolling along the underside of the $x$ axis.
Here $a = 1$; the slider runs $\theta$ from 0 to $\theta_{\max}$, drawing the curve arch after arch. The plot shows $-y$, so down is down.

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact

a = 1.0

def plot_cycloid(theta_max=np.pi):
    theta = np.linspace(0, theta_max, 1000)
    x = a * (theta - np.sin(theta))
    y = a * (1 - np.cos(theta))
    plt.figure(figsize=(12, 3))
    plt.plot(x, -y, 'k')
    plt.plot(x[-1], -y[-1], 'ro')       # where the curve has got to
    plt.plot(0, 0, 'bo')                # point 1, the start
    plt.xlim(0, 6 * np.pi * a)
    plt.ylim(-2.1 * a, 0.1 * a)
    plt.gca().set_aspect('equal')
    plt.title(fr'$\theta_{{\max}} = {theta_max / np.pi:.2f}\,\pi$')
    plt.xlabel('x'); plt.ylabel('-y (down is down)')
    plt.show()

interact(plot_cycloid, theta_max=(0.1, 6 * np.pi, 0.05));
```

## Interesting cases

- $\theta_{\max} = \pi$ (the default): down to the bottom of the first arch, $x = \pi a$, $y = 2a$ (point 3 in Taylor's Figure 6.5). The curve leaves point 1 going straight down.
- $\theta_{\max} = 2\pi$: one whole arch, back up to $y = 0$ at $x = 2\pi a$. Only one arch matters for the brachistochrone.
- $\theta_{\max} = 6\pi$: three arches, the cycloid repeating "indefinitely in a succession of loops" (Taylor, Problem 6.14), with a cusp at every $x = 2\pi n a$.
